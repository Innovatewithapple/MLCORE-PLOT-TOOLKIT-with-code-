```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
file_path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/GkDzb7bWrtvGXdPOfk6CIg/Obesity-level-prediction-dataset.csv"
data = pd.read_csv(file_path)
data.head()
```

🔹 1. Target only → countplot
👉 Use for:

Check class balance

```python
sns.countplot(y='NObeyesdad', data=data)
plt.title('Distribution of Target')
plt.show()
```

🔹 2. Numerical features → histogram
👉 Use for:

Distribution (normal / skewed)

```python
data[['Age','Weight','Height']].hist(figsize=(10,6))
plt.show()
```

🔹 3. Numerical → boxplot

👉 Use for:

Outliers detection

```python
sns.boxplot(y='Weight', data=data)
plt.show()
```

🔹 4. Numerical vs Target (VERY IMPORTANT)

👉 Use for:

See how feature affects target

```python
sns.boxplot(x='NObeyesdad', y='Weight', data=data)
plt.xticks(rotation=45)
plt.show()
```

🔹 5. Categorical vs Target

👉 Use for:

See category impact on target

```python
sns.countplot(x='Gender', hue='NObeyesdad', data=data)
plt.xticks(rotation=45)
plt.show()
```

🔹 6. Feature vs Feature (numerical)

👉 Use for:

Relationship between features

```python
sns.scatterplot(x='Weight', y='Height', data=data)
plt.show()
```

🔹 7. Correlation heatmap

👉 Use for:

Feature relationships + multicollinearity

```python
corr = data.corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.show()
```

🧠 SIMPLE CHEAT SHEET (REMEMBER THIS)
Target only → countplot  

Numeric → histogram + boxplot  

Categorical → countplot  

Feature vs target → boxplot / hue  

Feature vs feature → scatter / heatmap
